# Exploratory Data Analysis from Bitcoin Daily Prices

Exploratory Data Analysis (EDA)
- Visualize Price Trends: Plot Close price over time to see long-term trends, cycles, and major events.

- Consider a logarithmic scale for price to better visualize percentage changes and early price movements, given Bitcoin's exponential growth.

- Visualize Volume: Plot Volume over time to see trading activity. High volume often accompanies significant price movements.

- Correlation Matrix: See how Open, High, Low, Close, Volume (and your new features like returns, volatility) correlate with each other.

- Distribution Analysis: Histograms of returns, volume, etc., to understand their spread and common values.

- Seasonal Decomposition: Break down the time series into trend, seasonal, and residual components (e.g., using statsmodels.tsa.seasonal.seasonal_decompose). Bitcoin, while volatile, can exhibit some weekly or even daily patterns.

#### Library Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import seasonal_decompose
```

#### 1. Clean the Data

```python
# Load the data from the 'data' folder (one level up)
file_path = "../data/transformed_data/enriched_bitcoin_historical_price.csv"

# Create the data frame (df) object
df_bitcoin_enriched = pd.read_csv(file_path, index_col='Date', parse_dates=True) # Ensure 'Date' is parsed and set as index

df_bitcoin_enriched.head()
```

```python
df_bitcoin_enriched.info()
```

#### 2. Visualize Price Trends: Plot Close price over time to see long-term trends, cycles, and major events.


```python
# Closing Price (Linear Scale) Graphic
plt.figure(figsize=(15,7))
plt.plot(df_bitcoin_enriched.index, df_bitcoin_enriched['Close'], label='Bit coin Close Price')
plt.title('Bitcoin Close Price Over Time (Linear Scale)', fontsize=16)
plt.xlabel('Data', fontsize=12)
plt.ylabel('Price (USD)', fontsize=12)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

#### 3. Consider a logarithmic scale for price to better visualize percentage changes and early price movements, given Bitcoin's exponential growth.



```python
# Logarithm scale
plt.figure(figsize=(15,7))
plt.plot(df_bitcoin_enriched.index, df_bitcoin_enriched['Close'], label='Bit coin Close Price')
plt.yscale('log')
plt.title('Bitcoin Close Price Over Time (Logarithmic Scale)', fontsize=16)
plt.xlabel('Data', fontsize=12)
plt.ylabel('Price (USD, Log Scale)', fontsize=12)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Add Moving Average in the same price graphic
plt.figure(figsize=(15, 7))
plt.plot(df_bitcoin_enriched.index, df_bitcoin_enriched['Close'], label='Close Price', color='blue', alpha=0.7)
plt.plot(df_bitcoin_enriched.index, df_bitcoin_enriched['SMA_20'], label='SMA_20', color='red', linestyle='--')
plt.plot(df_bitcoin_enriched.index, df_bitcoin_enriched['SMA_50'], label='SMA_50', color='orange', linestyle='--')
plt.plot(df_bitcoin_enriched.index, df_bitcoin_enriched['SMA_200'], label='SMA_200', color='green', linestyle='--')
plt.title('Bitcoin Close Price with 20-day, 50-day and 200-day Simple Moving Averages', fontsize=16)
plt.xlabel('Data')
plt.ylabel('Price (USD)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

#### 4. Visualize Volume: Plot Volume over time to see trading activity. High volume often accompanies significant price movements.

```python
plt.figure(figsize=(15, 7))
plt.plot(df_bitcoin_enriched.index, df_bitcoin_enriched['Volume'], label='Trading Volume', color='orange', alpha=0.7)
plt.title('Bitcoin daily Trading Volume Over Time', fontsize=16)
plt.xlabel('Data')
plt.ylabel('Volume')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

#### 5. Correlation Matrix: See how Open, High, Low, Close, Volume (and your new features like returns, volatility) correlate with each other.

```python
columns_for_corr = ['Open', 'High', 'Low', 'Close', 'Volume', 
                    'Simple_Returns', 'Log_Returns', 'SMA_20', 'RSI', 'MACD_Line', 
                    'Volatility_30D', 'Bollinger_Bands_Width']

df_bitcoin_enriched_corr = df_bitcoin_enriched[columns_for_corr].dropna()

plt.figure(figsize=(12, 10))
sns.heatmap(df_bitcoin_enriched_corr.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Correlation Matrix of Key Bitcoin Features', fontsize=16)
plt.show()
```

#### 6. Distribution Analysis: Histograms of returns, volume, etc., to understand their spread and common values.

```python
# Log Returns Histogram
plt.figure(figsize=(10, 6))
sns.histplot(df_bitcoin_enriched['Log_Returns'].dropna(), bins=100, kde=True, color='skyblue')
plt.title('Distribution of Bitcoin Daily Log Returns', fontsize=16)
plt.xlabel('Log Returns')
plt.ylabel('Frequency')
plt.grid(True)
plt.tight_layout()
plt.show()

# Volume Histogram
plt.figure(figsize=(10, 6))
sns.histplot(df_bitcoin_enriched['Volume'].dropna(), bins=50, kde=True, color='lightcoral')
plt.title('Distribution of Bitcoin Daily Volume', fontsize=16)
plt.xlabel('Volume')
plt.ylabel('Frequency')
plt.grid(True)
plt.tight_layout()
plt.show()

# RSI Histogram
plt.figure(figsize=(10, 6))
sns.histplot(df_bitcoin_enriched['RSI'].dropna(), bins=50, kde=True, color='lightcoral')
plt.title('Distribution of RSI', fontsize=16)
plt.xlabel('RSI Value')
plt.ylabel('Frequency')
plt.grid(True)
plt.tight_layout()
plt.show()
```



#### 7. Seasonal Decomposition: Break down the time series into trend, seasonal, and residual components (e.g., using statsmodels.tsa.seasonal.seasonal_decompose). Bitcoin, while volatile, can exhibit some weekly or even daily patterns.

```python
df_bitcoin_enriched_decomp = df_bitcoin_enriched['Close'].dropna()

try: 
    decomposition_weekly = seasonal_decompose(df_bitcoin_enriched_decomp, model='multiplicative', period=7) # 0 'additive'

    plt.figure(figsize=(12, 8))
    plt.subplot(411)
    plt.plot(decomposition_weekly.observed, label='Original')
    plt.legend(loc='best')
    plt.subplot(412)
    plt.plot(decomposition_weekly.trend, label='Trend')
    plt.legend(loc='best')
    plt.subplot(413)
    plt.plot(decomposition_weekly.seasonal, label='Seasonal (Weekly)')
    plt.legend(loc='best')
    plt.subplot(414)
    plt.plot(decomposition_weekly.resid, label='Residual')
    plt.legend(loc='best')
    plt.suptitle('Seasonal decomposition of Bitcoin Price (Weekly Seasonality)', y=1.02,  fontsize=16)
    plt.tight_layout()
    plt.show()
except Exception as e:
    print(f"We couldn't create the weekly decomposition: {e}. Make sure you're using enough continuous data")
    
```

### 4. Analysis of Metrics (SQL & Python)


- Key Performance Indicators (KPIs): As per your objective, calculate KPIs like:

    * Average Daily Return

    * Annualized Volatility

    * Max Drawdown: The largest peak-to-trough decline over a specific period.

    * Sharpe Ratio (if you consider risk-free rate): Measures risk-adjusted return.

    * Win Rate/Loss Rate (if simulating trading strategies)

    * Return on Investment (ROI) for specific periods.

- Trend Identification: Use moving averages to identify uptrends, downtrends, and sideways markets.

- Support and Resistance Levels: Identify price levels where the price tends to find "support" (stops falling) or "resistance" (stops rising).

### 5. Time Series Forecasting (Python)
This is a core part of your project, especially the "Proyecciones" and "Pronóstico (series de tiempo)" aspect.

- Statistical Models:

    * ARIMA/SARIMA: Autoregressive Integrated Moving Average (and Seasonal ARIMA for seasonal patterns).
    
    * Prophet (Facebook): Good for forecasting time series data with strong seasonal components and holidays.
    
- Machine Learning Models:

    * Regression Models: Linear Regression, Ridge, Lasso (using features you engineered).
    
    * Tree-based Models: Random Forest, Gradient Boosting (e.g., LightGBM, XGBoost).
    
    * Recurrent Neural Networks (RNNs), LSTMs: Especially powerful for sequential data like time series.

- Model Evaluation: Use metrics like Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), Mean Absolute Percentage Error (MAPE) to evaluate forecast accuracy.

- Backtesting: Simulate how your trading strategies (based on your forecasts or indicators) would have performed on historical data.
