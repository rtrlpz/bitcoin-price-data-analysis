### Data Extraction from Yahoo Finance website
- Link: https://finance.yahoo.com/quote/BTC-USD/

```python
# Imports 
import yfinance as yf
import pandas as pd
import os
```

```python
# Create Bitcoin ticker
ticker_bitcoin = "BTC-USD"
```

Note: A ticker is a symbol, a unique combination of letters and numbers that represents a particular stock or security listed on an exchange

```python
# Create a Ticker object
bitcoin = yf.Ticker(ticker_bitcoin)
```

#### Params to download historical data
1. Method: .history()
2. Params:
    - period: "1d", "1mo", "1y", "ytd", "max".
    - interval: "1m", "1h", "1d", "1wk", "1mo".
    - start: "YYYY-MM-DD".
    - end: "YYYY-MM-DD".

#### Download max historical data by 1-day interval:

```python
# Extract all the historical price data 
historical_data_max_daily_df = bitcoin.history(period="max", interval="1d")
historical_data_max_daily_df.info()
```

```python
# Use the .describe() method to get quick statistical insight from the data
historical_data_max_daily_df.describe()
```

```python
# Show the Dataframe
historical_data_max_daily_df.head()
```

### Save the data in the data folder

```python
# Create a variable that holds the file name
file_name = "bitcoin_historical_price_daily.csv"

# Create the alternative route to the "data" folder
route_folder = os.path.join("..", "data", "raw", file_name)

# Save the DataFrame in the specified route folder
historical_data_max_daily_df.to_csv(route_folder, index=True)

print(f"The file '{file_name}' was saved successfully in: {route_folder}")
```
